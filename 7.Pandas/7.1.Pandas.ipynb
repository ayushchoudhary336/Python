Pandas- Dataframes and Series

Pandas is a powerful data manipulation library in python,widely used for data analysis and data cleaning .It provides two primary data structures :series and DataFrame. A series is a one dimensional arry like object . while a dataframes is a two dimensional , size-mutable, and potentially heterogeneous tabular data structure with labeled axes(rows and columns)


```python
import pandas as pd
```

```python
## Series

data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
## create a Series from dictionary 

data={"a":1,"b":2,"c":3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']

pd.Series(data,index=index)
```

```python
## Dataframe
## create aDataframe from a dictionary of list

data={
  'name':['Ayush','rohit','Sanu'],
  'Age':[25,30,45],
  'city':['Delhi','New York','Florida']
}
df=pd.DataFrame(data)
print(df)
print(type(df))


```
