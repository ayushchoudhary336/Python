Pandas- Dataframes and Series

Pandas is a powerful data manipulation library in python,widely used for data analysis and data cleaning .It provides two primary data structures :series and DataFrame. A series is a one dimensional arry like object . while a dataframes is a two dimensional , size-mutable, and potentially heterogeneous tabular data structure with labeled axes(rows and columns)


```python
import pandas as pd
```

```python
## Series

data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
## create a Series from dictionary 

data={"a":1,"b":2,"c":3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']

pd.Series(data,index=index)
```

```python
## Data Aggregating and Grouping

grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)


```

```python
## calculate the sum of products of each region
grouped_sum=df.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum)
```

```python
## aggregate multiple fuctions
grouped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
grouped_agg
```

```python
### Merging and Joining DataFrames

### create sample data frame
df1= pd.DataFrame({'Key': ['A','B','C'], 'Value1': [1,2,3]})
df2= pd.DataFrame({'Key': ['A','B','D'], 'Value2': [4,5,6]})

df1

```

```python
df2
```

```python
##merge dataframes on the key columns

pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
