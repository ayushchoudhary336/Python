```python
## Read a whole file

with open('final.txt','r') as file:
  content=file.read()
  print(content)
```

```python
## writing a file(overwriting)

with open('sample.txt','w') as file:
  file.write("maze karo bhai\n")
  file.write("ye sab moh maya hai .")
```

```python
##write a file without overwriting

with open('sample.txt','a') as file:
  file.write("Append opeartions taking place")
```

```python
## writing a list of lines
lines=['first line\n','Second line\n','Third line\n']
with open('sample.txt','a') as file:
  file.writelines(lines)
```

```python
## Binary files

## writing a binary data

data= b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
  file.write(data)


```

```python
## read the content from the source text file and write to a adestination text file

with open('sample.txt','r') as source_file:
  content=source_file.read()

with open('destination.txt','w') as destination_files:
  destination_files.write(content)

```
