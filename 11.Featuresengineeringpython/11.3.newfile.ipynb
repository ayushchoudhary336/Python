```python
import numpy as np
import pandas as pd
from sklearn.utils import resample
```

```python
#set the random seeed for reproductivity ...sets a fixed starting point for NumPy's random number generator so that the same random numbers are produced every time the code runs.

np.random.seed(123)
```

```python
n_samples = 1000
n_class_0 = 900  # Majority
n_class_1 = 100
```

```python
# create dataframe with imbalanced dataset
##pd.DataFrame() creates a table-like data structure using Pandas.
##np.random.normal() generates random numbers from a normal distribution with specified mean (loc) and standard deviation (scale).
##The 'target' column is created to indicate the class labels, where class 0 is labeled as 1 and class 1 is labeled as 0. 

class_0 = pd.DataFrame({
    'feature1': np.random.normal(0, 1, n_class_0),
    'feature2': np.random.normal(0, 1, n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature1': np.random.normal(1, 1, n_class_1),
    'feature2': np.random.normal(1, 1, n_class_1),
    'target': [1] * n_class_1
})
```

```python
##pd.concat() is used to concatenate the two dataframes (class_0 and class_1) along the rows (axis=0). The reset_index(drop=True) is used to reset the index of the resulting dataframe, dropping the old index and creating a new one. This ensures that the index is sequential and does not retain the original indices from the concatenated dataframes.

df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
##value_counts() is a Pandas function that counts the occurrences of each unique value in a Series. In this case, it is used to count the occurrences of each class label in the 'target' column of the dataframe df. This will help us understand the distribution of classes in our imbalanced dataset.

df['target'].value_counts()
```

```python
# 3. Data ko split karein (Target ke basis par)
df_majority = df[df['target'] == 0]
df_minority = df[df['target'] == 1]
```

```python

##resample() is a function from the sklearn.utils module that allows you to perform resampling operations on your data. In this case, it is used to upsample the minority class (class 0) by randomly sampling with replacement from the minority class dataframe (df_minority) until we have the same number of samples as the majority class (n_class_1). The random_state parameter is set to 123 to ensure reproducibility of the results.


##The resampled minority class dataframe (df_minority_upsampled) will contain the same number of samples as the majority class, effectively balancing the dataset. The replace=True parameter allows for sampling with replacement, meaning that the same sample from the minority class can be selected multiple times to achieve the desired number of samples.


df_minority_upsampled = resample(df_minority, 
                                 replace=True,     # Duplicate rows banane ke liye
                                 n_samples=len(df_majority), # Isko 900 kar do
                                 random_state=42)
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])

```
