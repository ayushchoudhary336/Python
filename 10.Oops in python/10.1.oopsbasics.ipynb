OOPS in python

Python is an object-oriented language, allowing you to structure your code using classes and objects for better organization and reusability.

Provides a clear structure to programs
Makes code easier to maintain, reuse, and debug
Helps keep your code DRY (Don't Repeat Yourself)
Allows you to build reusable applications with less code


When you create an object from a class, it inherits all the variables and functions defined inside that class.


Why Python Doesn't Use new

Python handles memory allocation automatically behind the scenes.
When you write:

p1 = Myclass()

Python internally:
allocates memory
creates the object
calls **init**() constructor
So the new keyword is not needed.


```python
class Myclass:
  x=5

p1= Myclass()
print(p1.x)
```

```python
##You can delete objects by using the del keyword:
del p1
```

```python
#Multiple Objects

p1 = Myclass()
p2 = Myclass()
p3 = Myclass()

print(p1.x)
print(p2.x)
print(p3.x)

```

All classes have a built-in method called **init**(), which is always executed when the class is being initiated.

The **init**() method is used to assign values to object properties, or to perform operations that are necessary when the object is being created

Using **init**() makes it easier to create objects with initial values:


The self Parameter

The self parameter is a reference to the current instance of the class.

It is used to access properties and methods that belong to the class.


```python
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

p2 = Person("Ayush",21)    
print(p2.name)
print(p2.age)
```

```python
class Person:
  def __init__(self, name):
    self.name = name

  def greet(self):
    return "Hello, " + self.name

  def welcome(self):
    message = self.greet()
    print(message + "! Welcome to our website.")

p1 = Person("Ayush")
p1.welcome()
```

The **str**() Method
The **str**() method is a special method that controls what is returned when the object is printed:


```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} ({self.age})"

p1 = Person("Ayush", 22)
print(p1)
```
